```python
# Definimos el header con la API key que utilizaremos para hacer las consultas
API_key = "2a5mpah09fdj"
headers = {"X-eBirdApiToken": API_key}
```

```python
# Importaremos los datos de eBird a un archivo CSV. Eventualmente, se creará un dataframe de pandas a partir de este archivo CSV.

# Definimos los rangos de fechas
fechas = pd.date_range(start="2022-01-01", end="2025-07-31", freq="D")


# Definimos una función capaz de obtener los avistamientos diarios por región
def obtener_avistamientos_diarios(codigo_region, fecha):
    url = f"https://api.ebird.org/v2/data/obs/{codigo_region}/historic/{fecha.year}/{fecha.month}/{fecha.day}"
    response = requests.get(url, headers=headers)
    if response.status_code == 200:
        return response.json()
    else:
        print(f"Error al obtener datos: {response.status_code}")
        return []

# Iteramos sobre las fechas y regiones para recopilar los datos de avistamientos
avistamientos_data = []

for fecha in fechas:
    avistamiento = obtener_avistamientos_diarios("CL", fecha)
    avistamientos_data.extend(avistamiento)

# Convertimos los datos recopilados en un DataFrame de pandas
df_avistamientos = pd.DataFrame(avistamientos_data)

# Guardamos el DataFrame en un archivo CSV
df_avistamientos.to_csv("avistamientos_ebird.csv", index=False)
```

```python
# Hacemos una copia del dataframe original
df = pd.read_csv("avistamientos_ebird.csv")
```

```python
# Obtengamos los nombres locales de las aves desde la API de eBird

# Definimos la URL para obtener la taxonomía en español (empleando el código de país "es_CL" para Chile y en formato JSON)
url = "https://api.ebird.org/v2/ref/taxonomy/ebird?locale=es_CL&fmt=json"

response = requests.get(url, headers=headers)
taxonomia_json = response.json()

# Creamos un mapa de códigos de especie a nombres comunes
mapa_especies = {item["speciesCode"]: item["comName"] for item in taxonomia_json}

# Mapeamos los nombres comunes en el dataframe
df["comName"] = df["speciesCode"].map(mapa_especies)
```

```python
# Limpiamos el dataframe eliminando filas con valores NaN en columnas críticas
df.dropna(subset=["speciesCode", "obsDt", "lat", "lng"], inplace=True)
```

### 3.2 Tranformación (1 punto)

**a) (0.5 pts)** Transforme la columna `obsDt` de string a un objeto de tipo `datetime` únicamente con año, mes y día (YYYY-MM-DD). Luego, agregue las siguientes nuevas columnas:
* `month` con únicamente el mes del avistamiento (puede dejar el mes como número entero de 1 a 12).
* `year-month` con el mes y año fecha en formato YYYY-MM

```python
# Transformar obsDt a datetime:
# Utlizamos "mixed" para manejar diferentes formatos de fecha, ya que hay entradas del tipo "2023-05-15 14:30" y "2023-05-15"
df["obsDt"] = pd.to_datetime(df["obsDt"], format="mixed", errors="coerce").dt.date

# Columna con solo el mes (número 1–12):
df["month"] = pd.to_datetime(df["obsDt"]).dt.month

# Columna con año-mes (YYYY-MM):
df["year-month"] = pd.to_datetime(df["obsDt"]).dt.to_period("M").astype(str)

# Reordenamos las columnas para mayor claridad
columnas = ["speciesCode", "comName", "sciName", "locId", "locName", "obsDt", "month", "year-month", "howMany", 
            "lat", "lng", "obsValid", "obsReviewed", "locationPrivate", "subId", "exoticCategory"]

df = df[columnas]
df.head(10)
```

**b) (0.5 pts)** Averigue cómo tranformar su DataFrame de Pandas en un GeoDataFrame de GeoPandas. Para esto, utilice las columnas `lat`y `lng` para generar los puntos de la nueva columna `geometry`. Para las visualizaciones siguientes utilizaremos el Coordinate Reference System (CRS) EPSG 4326 (también llamado [WGS 84](https://es.wikipedia.org/wiki/WGS84)). Asegúrese que sus coordenadas correspondan a este CRS al crear su GeoDataFrame.

```python
# Respuesta
geometria = gpd.points_from_xy(df["lng"], df["lat"])
gdf = gpd.GeoDataFrame(df, geometry=geometria, crs="EPSG:4326")

gdf.head(10)
print("CRS:", gdf.crs)
```

```python
import matplotlib.colors as mcolors
colors = ["yellow", "green", "yellow"]
positions = [0, 0.5, 1]

# Creamos un colormap cíclico para representar meses del año
custom_cmap = mcolors.LinearSegmentedColormap.from_list("cyclic_cmap", list(zip(positions, colors)))
```
